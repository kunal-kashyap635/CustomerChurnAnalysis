# Creating Database


```python
import mysql.connector as msq
import pandas as pd
```

```python
# MySQL connection
conn = msq.connect(
    host="localhost",
    user="root",
    port=3308,
    password="Kunal@2909",
    database="customer"
)

cursor = conn.cursor()
print("Database Loaded Successfully...")
```

```python
# 2. Read Excel Sheets

file_path = "customer_churn_data_raw.xlsx"

# Making data
customer_df = pd.read_excel(file_path,sheet_name="db_customer")

subscription_df = pd.read_excel(file_path,sheet_name="db_subscription")

support_df = pd.read_excel(file_path,sheet_name="db_support")

# little addition so NAN values also imported 
customer_df = customer_df.astype(object).where(pd.notna(customer_df),None)

subscription_df = subscription_df.astype(object).where(pd.notna(subscription_df),None)

support_df = support_df.astype(object).where(pd.notna(support_df),None)

# 3. Insert Customer Raw Data

customer_query = """
INSERT INTO customer
VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
"""

cursor.executemany(customer_query,customer_df.values.tolist())

print("Customer data inserted!")

# 4. Insert Subscription Raw Data

subscription_query = """
INSERT INTO subscription
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

cursor.executemany(subscription_query,subscription_df.values.tolist())

print("Subscription data inserted!")

# 5. Insert Support Raw Data

support_query = """
INSERT INTO support
VALUES (%s, %s, %s, %s, %s, %s)
"""

cursor.executemany(support_query,support_df.values.tolist())

print("Support data inserted!")

# 6. Commit

conn.commit()
print("All raw data inserted successfully!")

# 7. Close Connection

cursor.close()
conn.close()

print("MySQL connection closed!")
```
